Os conceitos de "distribuição" e "amostragem" são essenciais na quantificação e no entendimento da aleatoriedade dos dados. Nessa aula, vamos aprender a  como incluir a incerteza do processo de amostragem nos nossos resultados.

```python
# Imports para esse tópico
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('ggplot')

# Opções de como printar objetos do Numpy e do Pandas
np.set_printoptions(threshold = 20, precision = 2, suppress = True)
pd.set_option("display.max_rows", 7)
pd.set_option("display.max_columns", 8)
pd.set_option("display.precision", 2)
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Number of total trials/flips
num_trials = 10000

# Simulate fair coin flips: outcomes 0 or 1 with equal probability
flips = np.random.binomial(n=1, p=0.5, size=num_trials)

# Calculate cumulative sum and divide by the number of trials up to that point
running_average = np.cumsum(flips) / np.arange(1, num_trials + 1)

# Plotting the convergence
plt.figure(figsize=(12, 6))
plt.plot(running_average, color='royalblue', label='Sample Mean')
plt.axhline(0.5, color='red', linestyle='--', label='True Expected Value (μ = 0.5)')
plt.title('Weak Law of Large Numbers Demonstration', fontsize=14)
plt.xlabel('Number of Trials (n)', fontsize=12)
plt.ylabel('Sample Mean', fontsize=12)
plt.ylim(0.0, 1.0)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

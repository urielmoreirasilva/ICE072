```python
## Imports para esse tópico
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('ggplot')

## Opções de como printar objetos do Numpy e do Pandas
np.set_printoptions(threshold = 20, precision = 2, suppress = True)
pd.set_option("display.max_rows", 7)
pd.set_option("display.max_columns", 8)
pd.set_option("display.precision", 2)
```

## Exemplo #1: $N(0, 1)$

```python
"""
Exemplo #1: Y_i ~ iid. Normal(0, 1)
"""

## Semente aleatória (para garantir reprodutibilidade)
np.random.seed(42)

## Tamanho da sequência
n = int(1e3)

## Quantidade de sequências independentes
m = int(20)

## Parâmetros da simulação
mu = 0 # média populacional
sigma2 = 1 # variância populacional

## Declarando objetos
vY = np.zeros((n, m)) # v.a.'s
vY_bar = np.zeros((n, m)) # médias

## Loop principal
for j in np.arange(0, m, 1):
    for i in np.arange(0, n, 1):
        vY[i, j] = np.random.randn() 
        vY_bar[i, j] = vY[0:i+1, j].mean()
```

```python
## Plotting the convergence
plt.figure(figsize=(12, 6))
plt.plot(vY_bar, color='royalblue', label='Sample Mean')
plt.axhline(0, color='red', linestyle='--', label='True Expected Value (μ = 0)')
plt.title('Weak Law of Large Numbers Demonstration', fontsize=14)
plt.xlabel('Number of Trials (n)', fontsize=12)
plt.ylabel('Sample Mean', fontsize=12)
# plt.ylim(0.0, 1.0)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

## include epsilon! [!]
```

```python
## Calculando EQMs
vMSE = np.zeros(n) # EQMs

## Loop principal
for i in np.arange(0, n, 1):
    vMSE[i] = ((vY_bar[i, :] - mu)**2).mean()
```

```python
## Plotting the convergence
plt.figure(figsize=(12, 6))
plt.plot(vMSE, color='royalblue', label='Sample Mean')
plt.axhline(0, color='red', linestyle='--', label='True Expected Value (μ = 0)')
plt.title('Weak Law of Large Numbers Demonstration', fontsize=14)
plt.xlabel('Number of Trials (n)', fontsize=12)
plt.ylabel('Sample Mean', fontsize=12)
# plt.ylim(0.0, 1.0)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

## include epsilon! [!]
```

```python
## checando o resultado analítico! [!]

i = 10
print(vMSE[i])
print(1 / (i + 1))
```
